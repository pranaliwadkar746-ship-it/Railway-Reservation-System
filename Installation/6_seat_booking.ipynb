Task 1: Simulate Bookings Using While Loop



```python
total_seats = 5

while total_seats > 0:
    print ("seat Booked Successfully")
    print("Remaining seats:",total_seats - 1)
    total_seats -= 1

    print("All seats are Filled")
```
